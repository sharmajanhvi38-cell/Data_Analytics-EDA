```python
import pandas as pd
data ={
    "Student":["A","B","C","D","E","F","G","H","I","J"],
    "Study_Hours":[2,5,3,7,4,6,1,8,5,3],
    "Attendance":[70,85,75,95,80,90,65,98,88,72],
    "Marks":[55,78,62,90,70,85,45,95,82,60]
}
df = pd.DataFrame(data)
df
```

```python
print("Dataset Shape:",df.shape)
print("\nColumn Names:")
print(df.columns.tolist())
print("\nBasic Information:")
df.info()
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
print("Statistical Summary:")
df.describe()
```

```python
print("Study Hours vs Marks:")
print(df[["Study_Hours","Marks"]].sort_values("Study_Hours"))
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8,5))
plt.scatter(df["Study_Hours"],df["Marks"])
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.plot(df["Attendance"],df["Marks"])
plt.xlabel("Attendance(%)")
plt.ylabel("Marks")
plt.title("Attendance vs Marks")
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.bar(df["Student"],df["Marks"])
plt.xlabel("Student")
plt.ylabel("Marks")
plt.title("Marks of Comparision")
plt.show()
```

```python
plt.figure(figsize=(18,5))
plt.hist(df["Study_Hours"],bins=5)
plt.xlabel("Study Hours")
plt.ylabel("Number of Students")
plt.title(" Distribution of Salary")
plt.show( )
```

```python
correlation = df[["Study_Hours","Attendance","Marks"]].corr()
print("Correlation Matrix:")
print(correlation)
```

Students who spend more study hours generally achieve higher marks.
Higher attendance is generally associated with better marks.
The dataset contains no missing values.
Study hours and attendance show a strong relationship with marks.
